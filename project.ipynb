```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df = pd.read_csv("qoute_dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
quotes = df['quote']
quotes.head()
```

```python
quotes = quotes.str.lower()
quotes.head()
```

```python
import string
quotes = quotes.str.translate(str.maketrans('', '', string.punctuation))
quotes.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 10000

tokinizer = Tokenizer(num_words=vocab_size)
tokinizer.fit_on_texts(quotes)
```

```python
word_index = tokinizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python

sequence = tokinizer.texts_to_sequences(quotes)
```

```python
for i in range(3):
  print(quotes[i])
```

```python
for i in range(3):
  print(sequence[i])
```

```python
X = []
y = []

for seq in sequence:
  for i in range(len(seq)):
    input_seq = seq[:i]
    output_seq = seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
X
```

```python
len(X)
len(y)
```

```python
max_len = max(len(x) for x in X)
print(max_len)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X, maxlen=max_len, padding='pre')
```

```python
X_padded
```

```python
y = np.array(y)
```

```python
X_padded.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y,num_classes= vocab_size)
```

```python
y_one_hot.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN, LSTM, Dense
```

```python
embedding = 50
rnn_unit=128
```

```python
rnn_model = Sequential()
rnn_model.add(Embedding(input_dim=vocab_size, output_dim=embedding, input_length=max_len))

rnn_model.add(SimpleRNN(units=rnn_unit))

rnn_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
rnn_model.summary()
```

```python
lstm_model = Sequential()
lstm_model.add(Embedding(input_dim=vocab_size, output_dim=embedding, input_length=max_len))

lstm_model.add(LSTM(units=rnn_unit))

lstm_model.add(Dense(units=vocab_size, activation='softmax'))
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
lstm_model.summary()
```

```python
epochs = 100
batch_size =128
```

```python
history_rnn = rnn_model.fit(
    X_padded, y_one_hot,
    epochs=epochs,
    batch_size=batch_size,
    validation_split=0.1
)
```
